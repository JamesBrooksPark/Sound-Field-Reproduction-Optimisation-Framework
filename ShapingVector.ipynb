```python
from room_augmentation.functions.FunctionLibrary import FIRShaper

import matplotlib.pyplot as plt

I = 2**12

Delta = int(44100*0.01)
```

```python
x_1 = FIRShaper(I+Delta,I, 0, type='exponential')
x_2 = FIRShaper(I+Delta,I, Delta, type='exponential')
```

```python
fig, (axes1) = plt.subplots(1, 1,figsize=(8,4))

t = [x / 44100 for x in range(len(x_1))]

axes1.plot(t, x_1, color = 'black')
axes1.plot(t, x_2, color = 'grey')

axes1.set_ylabel("weighting", fontsize = 16)
axes1.set_xlabel("time(s)", fontsize = 16)

axes1.set_yscale('log')
plt.tight_layout()
# plt.show()


plt.savefig('./weighting_plot.pdf')
plt.show()
```
